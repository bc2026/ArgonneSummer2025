# SPECIFY C and M from Regression Equation
# SPECIFY FLOW_RATE

```python
c = 0
m = 103054
FLOW_RATE = 0.403
AMOUNT_OF_ELEMENT = 1.176
```

```python
from utils.DetectionTools import sanitize_response_name

clean_response = sanitize_response_name(response)

final_df = applied_voltage_df.copy()
final_df['Corrected_Time'] = (interp_df.index - (25 + 2730))

final_df[f'{clean_response}-ppb'] = (final_df[response] - c)/m
final_df[f'{clean_response}-ug/hr'] = final_df[f'{clean_response}-ppb'] * 0.001 * 60 * FLOW_RATE
final_df[f'{clean_response}-ug/hr/(ug-[{clean_response}])'] = final_df[f'{clean_response}-ug/hr'] / AMOUNT_OF_ELEMENT

# final_df.set_index('Time', inplace=True)
final_df.to_csv(r'../output/ICPMS-EChem-Combined.csv')
final_df
```
